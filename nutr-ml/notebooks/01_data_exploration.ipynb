# 01 — Dataset Exploration
Inspect image and text datasets before training.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from pathlib import Path

IMAGE_DIR = '../datasets/images'
TEXT_DIR  = '../datasets/texts'
```

## Image Dataset

```python
for split in ['train', 'val', 'test']:
    split_path = Path(IMAGE_DIR) / split
    if not split_path.exists():
        continue
    classes = [d.name for d in split_path.iterdir() if d.is_dir()]
    counts  = [len(list((split_path / c).glob('*.*'))) for c in classes]
    print(f'\n{split.upper()} — {len(classes)} classes, {sum(counts)} total images')
    for c, n in sorted(zip(classes, counts), key=lambda x: -x[1]):
        print(f'  {c}: {n}')
```

```python
# Show sample images from each class
train_path = Path(IMAGE_DIR) / 'train'
if train_path.exists():
    classes = sorted([d.name for d in train_path.iterdir() if d.is_dir()])[:8]
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    for ax, cls in zip(axes.flat, classes):
        imgs = list((train_path / cls).glob('*.jpg')) + list((train_path / cls).glob('*.png'))
        if imgs:
            ax.imshow(mpimg.imread(imgs[0]))
        ax.set_title(cls, fontsize=9)
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```
